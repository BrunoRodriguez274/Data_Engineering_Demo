# Create Payments

```python
import pyspark.sql.functions as F
import pyspark.sql.types as T
import datetime as dt
```

```python
payments_schema = T.StructType([
    T.StructField("Payment_Id", T.StringType(), False),
    T.StructField("Plan_Id", T.IntegerType(), False),
    T.StructField("Customer_Id", T.IntegerType(), False),
    T.StructField("Payment_Amount", T.FloatType(), False),
    T.StructField("Payment_Time", T.TimestampType(), False),
    T.StructField("Plan_Expiration_Time", T.TimestampType(), True),
    T.StructField("Payment_Status", T.StringType(), False),
    T.StructField("Updated_At", T.TimestampType(), False)
])

# access_polars_schema = {"Visit_Id" : pl.Int64,
#                     "Customer_Id" : pl.Int64,
#                     "Plan_Id" : pl.Int64,
#                     "Branch_Id" : pl.Int8,
#                     "Payment_Id" : pl.String,
#                     "Date" : pl.Date,
#                     "Visit_Time" : pl.Datetime,
#                     "Updated_At" : pl.Datetime
# }
```

```python
df_payment = spark.read.csv("/Volumes/momentum/bronze/external_files/payments_v2.csv", 
    header = True, 
    schema = payments_schema
)
```

```python
df_payment.display()
```

```python
df_payment.write.mode("overwrite").saveAsTable("momentum.bronze.payments")
```
